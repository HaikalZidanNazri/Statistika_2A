```python
import pandas as pd
from sklearn.model_selection import train_test_split
from google.colab import drive
drive.mount('/content/drive')

data = pd.read_csv('/content/drive/MyDrive/Statistika/MovieData-2018-2020.csv')

# cari genre yang hanya memiliki 1 anggota (tidak bisa distratifikasi)
genre_counts = data['genre'].value_counts()
valid_genres = genre_counts[genre_counts >= 2].index
data = data[data['genre'].isin(valid_genres)]

_, stratified_sample = train_test_split(
    data,
    test_size=0.3,           # Ambil 30% data dari setiap genre
    stratify=data['genre'],  # Stratifikasi berdasarkan kolom 'genre'
    random_state=17          # 3 digit akhir NIM: 017
)
print('Stratified Sampling:\n', stratified_sample)
```

```python
sample_30 = stratified_sample.head(30).reset_index(drop=True)
# 30 data teratas dari hasil sampling

_, stratified_sample = train_test_split(
    sample_30,
    test_size=0.3,              # Ambil 30% data dari setiap genre
    stratify=sample_30['genre'], # Stratifikasi berdasarkan kolom 'genre'
    random_state=27              # test_size sesuai nim 027 = 27
)
print('Stratified Sampling (20 data teratas):\n', sample_30)
```

```python
import scipy.stats as stats
import math

# menentukan "mean" sample (x_bar), sample standard deviasi (s), sample size (n)  confidence level
x_bar = sample_30['budget'].mean()
s = sample_30['budget'].std(ddof=1)
n = len(sample_30)
confidence_level = 0.95

# Menghitung alpha, degrees of freedom (df), critical t-value, dan margin of error
alpha = (1 - confidence_level)
df = n - 1
standard_error = s / math.sqrt(n)
critical_t = stats.t.ppf(1 - alpha/2, df)
margin_of_error = critical_t * standard_error

# menghitung batas atas dan batas bawah Confidence Interval
lower_bound = x_bar - margin_of_error
upper_bound = x_bar + margin_of_error

# Print hasil
print("Critical t-value: {:.3f}".format(critical_t))
print("Margin of Error: {:.3f}".format(margin_of_error))
print("Confidence Interval: [{:.3f},{:.3f}]".format(lower_bound, upper_bound))
print("The {:.1%} confidence interval dari populasi mean adalah:".format(confidence_level))
print("antara {:.3f} dan {:.3f}".format(lower_bound, upper_bound))
```
